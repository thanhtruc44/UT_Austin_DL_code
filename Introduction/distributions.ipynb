```python
import torch
import torch.distributions as dict
```

```python
# Let create a Bernoulli distribution
bernoulli = dict.Bernoulli(torch.tensor([0.5]))
print(bernoulli)
# Sample from the distribution
for _ in range(10):
    print(bernoulli.sample())
```

```python
sample = [bernoulli.sample() for _ in range(1000)]
print(f"Mean of the samples: {torch.mean(torch.tensor(sample))}")
print(f"Variance of the samples: {torch.var(torch.tensor(sample))}")
sample2 = [bernoulli.sample() for _ in range(10000)]
print(f"Mean of the samples: {torch.mean(torch.tensor(sample2))}")
print(f"Variance of the samples: {torch.var(torch.tensor(sample2))}")
```

```python
# Create a normal distribution
normal = dict.Normal(torch.tensor([0.0]), torch.tensor([1.0]))
for _ in range(10):
    print(normal.sample())
```

```python
sample = [normal.sample() for _ in range(1000)]
print(f"Mean of the samples: {torch.mean(torch.tensor(sample))}")
print(f"Variance of the samples: {torch.var(torch.tensor(sample))}")
sample2 = [normal.sample() for _ in range(10000)]
print(f"Mean of the samples: {torch.mean(torch.tensor(sample2))}")
print(f"Variance of the samples: {torch.var(torch.tensor(sample2))}")
```
