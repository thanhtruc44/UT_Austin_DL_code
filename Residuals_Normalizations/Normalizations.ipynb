```python
import torch
from typing import Any
```

```python
class MyModel(torch.nn.Module):
    def __init__(self, layer_size=[512, 512, 512]) -> None:
        super().__init__()
        layers = []
        layers.append(torch.nn.Flatten())
        c = 128 * 128 * 3
        for s in layer_size:
            layers.append(torch.nn.Linear(c, s))
            layers.append(torch.nn.ReLU())
            c = s
        layers.append(torch.nn.Linear(c, 102))
        self.model = torch.nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> Any:
        return self.model(x)
```

```python
x = torch.randn(10,3,128,128)
net0 = MyModel([])
print(net0(x).norm())
for n in range(10):
    netn = MyModel([512]*n)
    print(f"{n = } {netn(x).norm() = }")
```

```python
# Nobias

class MyModelNoBias(torch.nn.Module):
    def __init__(self, layer_size=[512, 512, 512]) -> None:
        super().__init__()
        layers = []
        layers.append(torch.nn.Flatten())
        c = 128 * 128 * 3
        for s in layer_size:
            layers.append(torch.nn.Linear(c, s, bias = False))
            layers.append(torch.nn.ReLU())
            c = s
        layers.append(torch.nn.Linear(c, 102, bias = False))
        self.model = torch.nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> Any:
        return self.model(x)
x = torch.randn(10,3,128,128)
for n in range(10):
    netn = MyModelNoBias([512]*n)
    print(f"{n = } {netn(x).norm() = }")
```

```python
# BatchNorm

class MyModelBN(torch.nn.Module):
    def __init__(self, layer_size=[512, 512, 512]) -> None:
        super().__init__()
        layers = []
        layers.append(torch.nn.Flatten())
        c = 128 * 128 * 3
        for s in layer_size:
            layers.append(torch.nn.Linear(c, s))
            layers.append(torch.nn.BatchNorm1d(s))
            layers.append(torch.nn.ReLU())
            c = s
        layers.append(torch.nn.Linear(c, 102))
        self.model = torch.nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> Any:
        return self.model(x)
x = torch.randn(10,3,128,128)
for n in range(10):
    netn = MyModelBN([512]*n)
    print(f"{n = } {netn(x).norm() = }")
```

```python
# LayerNorm 1 (Linear --> Norm)

class MyModelLN(torch.nn.Module):
    def __init__(self, layer_size=[512, 512, 512]) -> None:
        super().__init__()
        layers = []
        layers.append(torch.nn.Flatten())
        c = 128 * 128 * 3
        for s in layer_size:
            layers.append(torch.nn.Linear(c, s))
            layers.append(torch.nn.LayerNorm(s))
            layers.append(torch.nn.ReLU())
            c = s
        layers.append(torch.nn.Linear(c, 102))
        self.model = torch.nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> Any:
        return self.model(x)
x = torch.randn(10,3,128,128)
for n in range(10):
    netn = MyModelLN([512]*n)
    print(f"{n = } {netn(x).norm() = }")
```

```python
# LayerNorm 2 (Norm -> Linear)

class MyModelLN(torch.nn.Module):
    def __init__(self, layer_size=[512, 512, 512]) -> None:
        super().__init__()
        layers = []
        layers.append(torch.nn.Flatten())
        c = 128 * 128 * 3
        for s in layer_size:
            layers.append(torch.nn.LayerNorm(c, bias=False))
            layers.append(torch.nn.Linear(c, s))
            layers.append(torch.nn.ReLU())
            c = s
        layers.append(torch.nn.Linear(c, 102))
        self.model = torch.nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> Any:
        return self.model(x)
x = torch.randn(10,3,128,128)
for n in range(10):
    netn = MyModelLN([512]*n)
    print(f"{n = } {netn(x).norm() = }")
```
