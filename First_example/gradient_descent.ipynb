```python
import matplotlib.pyplot as plt
import torch

def f(x: torch.Tensor) -> torch.Tensor:
    return 3*x**2+torch.sin(5*x)

def grad_f(x: torch.Tensor) -> torch.Tensor:
    return 6*x+5*torch.cos(5*x)

plt.plot(torch.arange(-1,1, 0.01), f(torch.arange(-1,1,0.01)))
```

```python
x = torch.randn(1)
print(x)
for _ in range(100):
    x = x - 0.01*grad_f(x)
print(x)
    # plt.scatter(x, f(x), color='red')
```

```python
x = torch.randn(1)
all_x = []
for _ in range(100):
    x = x - 0.01*grad_f(x)
    all_x.append(x)

plt.plot(torch.arange(-1,1, 0.01), f(torch.arange(-1,1,0.01)))
plt.plot(torch.stack(all_x), f(torch.stack(all_x)), color='red')
print(all_x[0], all_x[-1])
```
