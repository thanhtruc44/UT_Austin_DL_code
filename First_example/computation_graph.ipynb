```python
y = torch.rand(10)
x = torch.rand(10, requires_grad=True)
x
```

```python
b = (x**2).mean() # Thõa điều kiện backward()
print(b)
b.backward()
```

```python
x = torch.tensor(2**20, requires_grad=True, device='cuda')
torch.cuda.memory_allocated()/1024/1024 #MB
```

```python
b = x
for _ in range(100):
    b = torch.relu(b)
torch.cuda.memory_allocated()/1024/1024 #MB
```

```python
b.sum().backward()
torch.cuda.memory_allocated()/1024/1024 #MB
```

```python
# CPU 
import torch

# 1. Khởi tạo tensor trên CPU với kiểu số thực và bật theo dõi đạo hàm
x = torch.tensor(2.0**20, dtype=torch.float32, requires_grad=True)
print("Dung lượng x (MB):", x.element_size() * x.nelement() / 1024 / 1024)

# 2. Cho qua vòng lặp ReLU 100 lần
b = x
for _ in range(100):
  b = torch.relu(b)

print("Dung lượng b (MB):", b.element_size() * b.nelement() / 1024 / 1024)

# 3. Tính tổng (để biến b thành scalar) và gọi backward
b.sum().backward()

# 4. Kiểm tra kết quả đạo hàm (.grad) thay vì gọi lệnh CUDA
print("Gradient của x:", x.grad)
```
